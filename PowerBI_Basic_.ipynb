Power BI Basics

-  Q1. What is Power BI and why is it used in businesses?

  --> Answer: Power BI is a business intelligence
    tool developed by Microsoft. It helps users convert raw data into meaningful reports, charts, and interactive dashboards.
    Businesses use Power BI to,
	  Analyze business performance.
    Track sales, expenses, and profits.
    Identify trends and patterns.
    Make better decisions using data.
    Share reports with team members but we have to take pro license for this.

 - Q2. Name and explain the three main components of Power BI.

->Answer:  1.	Power BI Desktop: A Windows application used to import, clean, transform, and analyze data and create reports.

2.	Power BI Service: An online platform used to publish, share, and collaborate on reports and dashboards.
3.	Power BI Mobile: A mobile application used to view and interact with reports and dashboards on smartphones and tablets.

- Q3. Explain the Power BI workflow.

-> Answer: The Power BI workflow consists steps
  
  >  Get Data
Import data from Excel, CSV, databases

2. Clean and Transform Data
Remove errors, duplicates, and unwanted values.
3. Model Data
Create relationships between tables.
4. Create Reports
Build charts, tables, and dashboards.
5. Publish and Share
Publish reports to Power BI Service.

- Q5. Write step-by-step instructions to load the Real_Estate dataset.

  -- Answer:
  1.	Open Power BI Desktop.
  2.	Click Home  Get Data.
  3.	Select the appropriate file type
  4.	Browse and select the Real_Estate dataset.
  5.	Click Open
  6.	Select the required table or sheet in the Navigator window.
  7.	Click Transform Data if you need to clean or modify the dataset.
  8.	Check the column names and data types.
  9.	Click Close & Apply to load the data into Power BI.

- Q6. Define Data View, Report View, and Model View.

-- Answer: Data View	- Displays the rows and      columns of loaded data. It is useful for checking values and creating calculated columns.

Report View	Used to create visualizations such as charts, graphs, tables, and interactive reports.

Model View	Displays tables and their relationships. It helps organize data and connect related tables.
